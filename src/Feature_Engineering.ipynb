```python
import pandas as pd
```

```python
df = pd.read_csv("D:\clg_project\AI-Financial-Wellness-Coach\data\processed\Master_indian_financial_transactions.csv")
df.sample(5)
```

```python
df.dtypes
```

```python
df["transaction_date"] = pd.to_datetime(df["transaction_date"])
```

```python
df["day"] = df["transaction_date"].dt.day
df.columns
```

```python
df["week"] = df["transaction_date"].dt.isocalendar().week
```

```python
df["month"] = df["transaction_date"].dt.month
```

```python
df.columns
```

```python
df["weekend"] = df["transaction_date"].dt.weekday.isin([5,6])
df["weekend"] = df["weekend"].astype(int)
```

```python
df["merchant_freq"] = df["merchant"].map(df["merchant"].value_counts())
```

```python
df.head(3)
```

```python
df.drop(columns="Unnamed: 0", inplace=True)
```

```python
df["avg_expense"] = df["amount"].mean()
```

```python
df["rolling_avg"] = df["amount"].rolling(window=3, min_periods=1).mean()
```

```python
df.sample(7)
```

```python
df["month_end"] = df["transaction_date"].dt.is_month_end.astype(int)
```

```python
df["salary_week"] = ((df["day"] >= 25) | (df["day"] <=2)).astype(int)
```

```python
fest_months = [10, 11, 12]
df["festival_season"] = df["transaction_date"].dt.month.isin(fest_months).astype(int)
```

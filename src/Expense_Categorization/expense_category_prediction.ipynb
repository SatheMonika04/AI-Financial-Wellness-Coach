## Train / Validation / Test Split

### Now we separate the data into three datasets:

70% → Training,
15% → Validation,
15% → Test

```python
feature_columns = [
    "transaction_text",
    "amount",
    "amount_log",
    "payment_method",
    "transaction_type"
]

target_column = "category"

X = model_df[feature_columns].copy()
y = model_df[target_column].copy()
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

### First split: separate Test set

```python
# Use train_test_split

from sklearn.model_selection import train_test_split

X_temp, X_test, y_temp, y_test = train_test_split(
    X,
    y,
    test_size=0.15,
    stratify=y,
    random_state=42
)

# X_test = 15%
# y_test = 15%

# X_temp = 85%
# y_temp = 85%

# Test = 1,050
# Remaining = 5,950
```

### Second split: Training and Validation

We want validation to ultimately be 15% of the entire dataset.

Currently we have 85% remaining.

Therefore:

15 / 85 = 0.17647

So we take approximately 17.65% of the remaining data for validation.

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_temp,
    y_temp,
    test_size=0.17647,
    stratify=y_temp,
    random_state=42
)

# Now approximately:

'''Train → 70%
Validation → 15%
Test → 15%'''
```

### Verify the Shapes

```python
print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

### Verify Total Rows

```python
print(
    len(X_train)
    + len(X_val)
    + len(X_test)
)

print(
    len(y_train)
    + len(y_val)
    + len(y_test)
)
```

### Verify Category Distribution

```python
print("TRAIN")
print(y_train.value_counts(normalize=True).mul(100).round(2))

print("\nVALIDATION")
print(y_val.value_counts(normalize=True).mul(100).round(2))

print("\nTEST")
print(y_test.value_counts(normalize=True).mul(100).round(2))
```

### Check the Smallest Class

```python
print("Investment counts:")

print(
    "Train:",
    (y_train == "Investment").sum()
)

print(
    "Validation:",
    (y_val == "Investment").sum()
)

print(
    "Test:",
    (y_test == "Investment").sum()
)
```

### Check for Data Leakage

```python
train_rows = set(
    X_train.astype(str).agg("|".join, axis=1)
)

val_rows = set(
    X_val.astype(str).agg("|".join, axis=1)
)

test_rows = set(
    X_test.astype(str).agg("|".join, axis=1)
)

print("Train ∩ Validation:", len(train_rows & val_rows))
print("Train ∩ Test:", len(train_rows & test_rows))
print("Validation ∩ Test:", len(val_rows & test_rows))
```

### Confirming whether these overlaps have the same category

```python
# Add target back
train_check = X_train.copy()
train_check["category"] = y_train.values

val_check = X_val.copy()
val_check["category"] = y_val.values

test_check = X_test.copy()
test_check["category"] = y_test.values


# Create complete row keys INCLUDING category
train_full_rows = set(
    train_check.astype(str).agg("|".join, axis=1)
)

val_full_rows = set(
    val_check.astype(str).agg("|".join, axis=1)
)

test_full_rows = set(
    test_check.astype(str).agg("|".join, axis=1)
)

print(
    "Complete rows Train ∩ Validation:",
    len(train_full_rows & val_full_rows)
)

print(
    "Complete rows Train ∩ Test:",
    len(train_full_rows & test_full_rows)
)

print(
    "Complete rows Validation ∩ Test:",
    len(val_full_rows & test_full_rows)
)
```

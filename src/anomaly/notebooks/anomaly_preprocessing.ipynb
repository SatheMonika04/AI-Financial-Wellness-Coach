### Data Cleaning & Datatype Correction.

```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
RAW_PATH = Path(
    r"D:\clg_project\AI-Financial-Wellness-Coach\data\anomaly\anomaly_transactions_raw.csv"
)

CLEAN_PATH = Path(
    r"D:\clg_project\AI-Financial-Wellness-Coach\data\anomaly\anomaly_transactions_cleaned.csv"
)

print("Raw file exists:", RAW_PATH.exists())
print("Cleaned file will be saved to:", CLEAN_PATH)
```

```python
# Load the raw anomaly dataset

df = pd.read_csv(RAW_PATH)

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
df_clean = df.copy()
```

### Missing Values

```python
missing = df_clean.isnull().sum()

print(missing)

print(
    "\nTotal missing values:",
    df_clean.isnull().sum().sum()
)
```

### Date Column cleaning

```python
df_clean["transaction_date"] = pd.to_datetime(
    df_clean["transaction_date"],
    errors="coerce"
)
print(df_clean["transaction_date"].dtype)
```

```python
## Checking for invalid transaction date
invalid_dates = df_clean["transaction_date"].isna().sum()

print("Invalid transaction dates:", invalid_dates)
```

```python
## Teansaction date range
print("Earliest date:", df_clean["transaction_date"].min())
print("Latest date:", df_clean["transaction_date"].max())
```

### Transaction Time Dataset

```python
time_test = pd.to_datetime(
    df_clean["transaction_time"],
    format="%H:%M:%S",
    errors="coerce"
)

print("Invalid transaction times:", time_test.isna().sum())
```

### Amount Column 

```python
print("Amount dtype:", df_clean["amount"].dtype)
```

```python
print("Missing amounts:", df_clean["amount"].isna().sum())
print("Negative amounts:", (df_clean["amount"] < 0).sum())
print("Zero amounts:", (df_clean["amount"] == 0).sum())
```

### Validating Transaction type

```python
print(df_clean["transaction_type"].value_counts())
print("\nUnique transaction types:")
print(df_clean["transaction_type"].unique())
```

### Validating categorical columns

```python
categorical_columns = [
    "currency",
    "payment_method",
    "transaction_type",
    "category"
]

for col in categorical_columns:
    print("\n" + "=" * 50)
    print(col)
    print("Unique values:", df_clean[col].nunique())
    print(df_clean[col].unique())
```

### Validating location columns

```python
location_columns = [
    "city",
    "state",
    "country"
]

for col in location_columns:
    print("\n" + "=" * 50)
    print(col)
    print("Unique values:", df_clean[col].nunique())
    print(df_clean[col].value_counts().head(20))
```

### Validatind temporal columns

```python
temporal_columns = [
    "day",
    "week",
    "month",
    "weekend",
    "month_end",
    "salary_week",
    "festival_season"
]

print(df_clean[temporal_columns].dtypes)
```

```python
derived_day = df_clean["transaction_date"].dt.day
derived_month = df_clean["transaction_date"].dt.month

print(
    "Day mismatches:",
    (df_clean["day"] != derived_day).sum()
)

print(
    "Month mismatches:",
    (df_clean["month"] != derived_month).sum()
)
```

### Checking Transaction ID

```python
print("Missing transaction IDs:",
      df_clean["transaction_id"].isna().sum())

print("Duplicate transaction IDs:",
      df_clean["transaction_id"].duplicated().sum())
```

### Final Check

```python
print(df_clean.dtypes)
```

```python
print("Rows:", len(df_clean))
print("Columns:", len(df_clean.columns))
print("Missing values:", df_clean.isna().sum().sum())
print("Duplicate rows:", df_clean.duplicated().sum())
print("Duplicate transaction IDs:",
      df_clean["transaction_id"].duplicated().sum())
```

## Saving the clean dataset

```python
df_clean.to_csv(
    CLEAN_PATH,
    index=False
)

print("Cleaned dataset saved successfully!")
print("Saved at:", CLEAN_PATH)
```
